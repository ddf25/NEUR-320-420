Below is a calculation of the change in Na+ concentration that would be expected to occur in a neuron in response to an amount of Na+ entry that would raise the membrane potential by 100 mV, which is a typical AP amplitude. The result shows that for a cell with radius = 10 um, the change in [Na+]i resulting from the number of Na+ ions that would produce a 100 mV depolarization is a ~0.03 % increase in [Na+]i. Thus, although ion movement through ion channels is expected to change the intracellular ion concentration, the effect is so small that it is usually ignored. Note that there are conditions under which ion entry can produce sizable changes in ion concentraion: (a) in small cells, which you can demonstrate by changing the cell radius at the top of the program and re-runing the program by clicking on the black dot with the arrow on the left, and (b) if multiple APs occur in close succession, leading to buildup of intracellular Na+.

```python
import math

#Cell Radius (in cm):10 um=10e-4  cm
r = 10e-04

#Cell Volume (cm^3) and Surface Area (cm^2)
V_cm3 = (4/3) * math.pi * (r**3)
A_cm2 =4 *math.pi * (r**2)

# Convert Volume from cm^3 to Liters (1 cm^3 = 1e-3 L)
V_liters = V_cm3 * 1e-03

# Specific Membrane Capacitance (Farads/cm^2) - typically ~1 uF/cm^2
Unit_Cap = 1e-06
C = A_cm2 * Unit_Cap

# Action potential Amplitude (100 mV = 0.1 V)
AP_amp = 100 * 1e-03

# Charge required (Coulombs)
Q = AP_amp * C

# Number of elementary charges (ions)
e_charge = 1.60217663e-19
N_ions = Q / e_charge
# Convert ions to moles using Avogadro's number
N_A = 6.02214076e23
moles_Na = N_ions / N_A

# Concentration increment in Molar (moles / Liter)
Increment_M = moles_Na / V_liters

# Initial Na+ Concentration in Molar (10 mM = 10 * 1e-3 M = 0.01 M)
C_init = 10 * 1e-03

# Percent Change
Percent_Change = (Increment_M / C_init) * 100

print(f"Volume: {V_liters:.3e} L")
print(f"Charge: {Q:.3e} C")
print(f"Initial Concentration: {C_init/1e-03:} mM")
print(f"Na+ Concentration Increment: {Increment_M * 1e3:.4f} mM")
print(f"Percent Change: {Percent_Change:.4f}%")
```
